```python
# Create a dictionary
student = {
    "name": "Rahul",
    "age": 20,
    "college": "MIT-WPU"
}

# Add
student["city"] = "Mumbai"
print("\nAfter Adding:", student)

# Update
student["age"] = 21
print("\nAfter Updating:", student)

# Delete
del student["college"]
print("\nAfter Deleting:", student)
```
